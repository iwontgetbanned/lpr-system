```python
import time
import torch
from ultralytics import YOLO
```

```python
print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
from pathlib import Path
from ultralytics import YOLO

footage = Path("footage")
models = ["yolov8n.pt", "yolo11n.pt", "yolo11s.pt", "yolo26n.pt", "yolo11m.pt"]

for m in models:
    model = YOLO(m)
    times, counts = [], {}
    for r in model.predict(footage, device=0, stream=True, save=True,
                           name=f"cmp_{Path(m).stem}", verbose=False):
        times.append(r.speed["inference"])
        for c in r.boxes.cls.tolist():
            name = r.names[int(c)]
            counts[name] = counts.get(name, 0) + 1
    avg = sum(times[5:]) / max(len(times[5:]), 1)  # skip warmup frames
    top = sorted(counts.items(), key=lambda x: -x[1])[:6]
    print(f"{m:12s} frames={len(times):5d}  avg inference={avg:5.1f} ms  (~{1000/avg:4.0f} FPS)  top={top}")
```

```python
model = YOLO("yolo11n.pt")
model.train(
    data=r"C:\Capstone\License-Plate-Data\data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    patience=10,
    name="plate_yolo11n",
)
```
