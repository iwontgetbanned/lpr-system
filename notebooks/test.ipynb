```python
from pathlib import Path
from ultralytics import YOLO

footage = Path("footage")
models = ["yolov8n.pt", "yolo11n.pt", "yolo11s.pt", "yolo26n.pt", "yolo11m.pt"]

for m in models:
    model = YOLO(m)
    times, counts = [], {}
    for r in model.predict(footage, device=0, stream=True, save=True,
                           name=f"cmp_{Path(m).stem}", verbose=False):
        times.append(r.speed["inference"])
        for c in r.boxes.cls.tolist():
            name = r.names[int(c)]
            counts[name] = counts.get(name, 0) + 1
    avg = sum(times[5:]) / max(len(times[5:]), 1)  # skip warmup frames
    top = sorted(counts.items(), key=lambda x: -x[1])[:6]
    print(f"{m:12s} frames={len(times):5d}  avg inference={avg:5.1f} ms  (~{1000/avg:4.0f} FPS)  top={top}")
```

Run OCR every fifth frame of that id, then print the text that appears most often when the vehicle leaves the shot. tracking so each car is read a few times

```python
from collections import Counter
from pathlib import Path
from ultralytics import YOLO
from fast_plate_ocr import LicensePlateRecognizer

video = r"C:\Capstone\lpr-system\data\samples\IMG_1381.MOV"
detector = YOLO(r"C:\Capstone\lpr-system\models\plate_detector.pt")
reader = LicensePlateRecognizer("cct-s-v2-global-model")
votes = {}
seen = {}

for result in detector.track(video, persist=True, tracker="bytetrack.yaml", device=0, imgsz=640, conf=0.25, stream=True, verbose=False):
    if result.boxes is None or result.boxes.id is None:
        continue
    frame = result.orig_img
    for box, track_id in zip(result.boxes.xyxy.cpu().tolist(), result.boxes.id.int().cpu().tolist()):
        seen[track_id] = seen.get(track_id, 0) + 1
        if seen[track_id] % 5 != 1:
            continue
        x1, y1, x2, y2 = (int(v) for v in box)
        text = reader.run(frame[y1:y2, x1:x2])[0].plate
        votes.setdefault(track_id, []).append(text)

for track_id, readings in votes.items():
    plate, count = Counter(readings).most_common(1)[0]
    print(f"track {track_id}: {plate} ({count}/{len(readings)})")
```
